## 1) Descripción del problema **[1 pts]**
Una plataforma de streaming está analizando el crecimiento de nuevos suscriptores.
Se ha observado que el número de nuevos suscriptores en el mes n sigue el siguiente
comportamiento:
* En el mes 0 se registraron 0 nuevos usuarios.
* En el mes 1 se registró 1 nuevo usuario.
* A partir del mes 2, el número de nuevos usuarios es igual a la suma de los nuevos usuarios de los dos meses anteriores, debido al efecto combinado de referidos del mes anterior y recomendaciones acumuladas del mes previo.
### Relación de recurrencia:

T(n) = 0 si n = 0

T(n) = 1 si n = 1

T(n) = T(n-1) + T(n-2) si n ≥ 2

Determinar con memoización cuántos nuevos suscriptores habrá en el mes n.

## 6) Código funcional **[15 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
memoria = {}

def suscriptores (n):
    if n in memoria:
        return memoria[n]

    if n == 0:
        resultado = 0

    elif n == 1:
        resultado = 1

    else:
        resultado = suscriptores(n-1)+suscriptores(n-2)

    memoria[n] = resultado

    return resultado



def main():
    n=int(input("Ingrese el mes: "))
    print("Nuevos suscriptores",suscriptores(n))

main()
```
