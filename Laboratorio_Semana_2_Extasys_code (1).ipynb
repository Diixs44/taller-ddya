## Enunciado

# Diseño
### Estrategia general de solución

1. Solicitar al usuario la cantidad de productos (n).

2. Registrar los n precios en una lista.

3. Solicitar el tipo de ordenamiento (1 para ascendente o 2 para descendente).

4. Aplicar el algoritmo Insertion Sort sobre la lista de precios según la opción elegida.

5. Mostrar la lista de precios ordenada.

### Tipo de algoritmo

* Algoritmo de ordenamiento: Insertion Sort.

* Técnica: Incremental, ya que construye la lista ordenada insertando un elemento a la vez en su posición correcta.

### Tipo de implementación

* Iterativa, porque utiliza ciclos (for y/o while) para recorrer e insertar los elementos.

### Incremental

Sí. En cada iteración se toma un elemento de la lista y se inserta en la posición adecuada dentro de la parte que ya está ordenada.

### Invariantes importantes

* Antes de cada iteración, los elementos desde la posición 0 hasta i-1 permanecen ordenados.

* Al finalizar el algoritmo, toda la lista queda ordenada según el criterio seleccionado (ascendente o descendente).

### Justificación

Insertion Sort es adecuado para este problema porque permite ordenar los precios de forma sencilla y eficiente para listas pequeñas o medianas. Además, cumple con el requisito del ejercicio al organizar los datos tanto en orden ascendente como descendente mediante pequeñas inserciones sucesivas.

### Complejidad:

#### Mejor caso
Condición: La lista ya está ordenada.
Ejemplo: [1000, 2000, 3000, 4000]
tᵢ = 1
T(n) = n - 1
Complejidad: O(n)

#### Peor caso
Condición: La lista está completamente invertida.
Ejemplo: [5000, 4000, 3000, 2000]
tᵢ = i
T(n) = 1 + 2 + 3 + ... + (n - 1)
T(n) = n(n - 1) / 2
T(n) = (n² - n) / 2
Complejidad: O(n²)

# Codigo



```python
# Solicitar la cantidad de productos
n = int(input("Ingrese la cantidad de productos: "))

# Registrar los precios
precios = []

for i in range(n):
    precio = float(input(f"Ingrese el precio del producto {i + 1}: "))
    precios.append(precio)

# Elegir el tipo de ordenamiento
print("\nTipo de ordenamiento:")
print("1. Ascendente")
print("2. Descendente")
opcion = int(input("Seleccione una opción: "))

# Insertion Sort
for i in range(1, len(precios)):
    clave = precios[i]
    j = i - 1

    if opcion == 1:  # Ascendente
        while j >= 0 and precios[j] > clave:
            precios[j + 1] = precios[j]
            j -= 1

    elif opcion == 2:  # Descendente
        while j >= 0 and precios[j] < clave:
            precios[j + 1] = precios[j]
            j -= 125

    else:
        print("Opción no válida.")
        break

    precios[j + 1] = clave

# Mostrar resultado
if opcion == 1:
    print("\nPrecios ordenados de menor a mayor:")
    print(precios)
elif opcion == 2:
    print("\nPrecios ordenados de mayor a menor:")
    print(precios)
```
